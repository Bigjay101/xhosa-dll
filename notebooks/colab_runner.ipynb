Mounts Drive, clones the repo, installs requirements and links `data/` and `results/` to the shared Drive folder so long runs survive disconnects.

```
from google.colab import drive
drive.mount('/content/drive')
```

```
REPO = 'https://github.com/<owner>/nlp-p2-isixhosa.git'   # private: use a token or gh auth
SHARED = '/content/drive/MyDrive/nlp-p2-shared'
!git clone -q $REPO /content/nlp-p2-isixhosa
%cd /content/nlp-p2-isixhosa
!pip -q install -r requirements.txt
```

```
import os
for name in ('data', 'results'):
    target = f'{SHARED}/{name}'
    os.makedirs(target, exist_ok=True)
    if os.path.islink(name) or not os.path.exists(name):
        if os.path.islink(name): os.unlink(name)
        os.symlink(target, name)
    else:
        print(f'{name}/ exists in the repo checkout; move it into {target} first')
!ls -la data results
```
